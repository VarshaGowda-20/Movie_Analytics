```python
df = movies.merge(
    credits,
    left_on="id",
    right_on="movie_id",
    how="left"
)

print("Merged Dataset Shape:")
print(df.shape)

display(df.head())
```

```python
df = df.drop(
    columns=["movie_id", "title_y"],
    errors="ignore"
)

df = df.rename(
    columns={"title_x": "title"}
)

print(df.shape)
print(df.columns.tolist())
```

```python
selected_columns = [
    "id",
    "title",
    "genres",
    "budget",
    "revenue",
    "release_date",
    "runtime",
    "popularity",
    "vote_average",
    "vote_count",
    "original_language",
    "production_companies",
    "cast",
    "crew"
]

df = df[
    [col for col in selected_columns if col in df.columns]
]

display(df.head())
```

```python
df["release_date"] = pd.to_datetime(
    df["release_date"],
    errors="coerce"
)

df["year"] = df["release_date"].dt.year

print(df[["title", "release_date", "year"]].head())
```

```python
numeric_columns = [
    "budget",
    "revenue",
    "runtime",
    "popularity",
    "vote_average",
    "vote_count"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

print(df[numeric_columns].dtypes)
```

```python
print("Budget = 0:", (df["budget"] == 0).sum())
print("Revenue = 0:", (df["revenue"] == 0).sum())
```

```python
df["profit"] = df["revenue"] - df["budget"]

display(
    df[
        ["title", "budget", "revenue", "profit"]
    ].head(10)
)
```

```python
df["roi"] = np.where(
    df["budget"] > 0,
    (
        (df["revenue"] - df["budget"])
        / df["budget"]
    ) * 100,
    np.nan
)

display(
    df[
        ["title", "budget", "revenue", "roi"]
    ].head(10)
)
```

```python
df["rating_category"] = pd.cut(
    df["vote_average"],
    bins=[
        -np.inf,
        5,
        7,
        8,
        np.inf
    ],
    labels=[
        "Low",
        "Average",
        "Good",
        "Excellent"
    ]
)

display(
    df[
        ["title", "vote_average", "rating_category"]
    ].head(10)
)
```

```python
def extract_names(value):
    try:
        data = ast.literal_eval(value)

        if isinstance(data, list):
            return [
                item["name"]
                for item in data
                if isinstance(item, dict)
                and "name" in item
            ]

    except (ValueError, SyntaxError, TypeError):
        return []

    return []


df["genre_list"] = df["genres"].apply(
    extract_names
)

df["genre"] = df["genre_list"].apply(
    lambda x: ", ".join(x)
)

display(
    df[
        ["title", "genre"]
    ].head(10)
)
```

```python
genre_df = df[
    ["id", "title", "genre_list", "vote_average",
     "budget", "revenue", "profit", "roi"]
].copy()

genre_df = genre_df.explode(
    "genre_list"
)

genre_df = genre_df.rename(
    columns={"genre_list": "genre"}
)

genre_df = genre_df.dropna(
    subset=["genre"]
)

display(genre_df.head(10))
```

```python
total_movies = df["id"].nunique()

average_rating = df["vote_average"].mean()

total_revenue = df["revenue"].sum()

average_revenue = df["revenue"].mean()

average_budget = df["budget"].mean()

average_roi = df["roi"].mean()

print("========== MOVIE ANALYTICS KPIs ==========")

print(f"Total Movies       : {total_movies:,}")
print(f"Average Rating     : {average_rating:.2f}")
print(f"Total Revenue      : {total_revenue:,.0f}")
print(f"Average Revenue    : {average_revenue:,.0f}")
print(f"Average Budget     : {average_budget:,.0f}")
print(f"Average ROI        : {average_roi:.2f}%")
```

```python
movies_by_genre = (
    genre_df["genre"]
    .value_counts()
)

print(movies_by_genre)
```

```python
rating_by_genre = (
    genre_df
    .groupby("genre")["vote_average"]
    .mean()
    .sort_values(
        ascending=False
    )
)

display(
    rating_by_genre
    .round(2)
    .to_frame("Average Rating")
)
```

```python
revenue_by_genre = (
    genre_df
    .groupby("genre")["revenue"]
    .mean()
    .sort_values(
        ascending=False
    )
)

display(
    revenue_by_genre
    .to_frame("Average Revenue")
)
```

```python
roi_by_genre = (
    genre_df
    .groupby("genre")["roi"]
    .mean()
    .sort_values(
        ascending=False
    )
)

display(
    roi_by_genre
    .round(2)
    .to_frame("Average ROI %")
)
```

```python
movies_by_year = (
    df
    .groupby("year")
    .size()
    .reset_index(
        name="movie_count"
    )
)

display(
    movies_by_year.tail(20)
)
```

```python
revenue_by_year = (
    df
    .groupby("year")["revenue"]
    .sum()
    .reset_index()
)

display(
    revenue_by_year
    .sort_values(
        "revenue",
        ascending=False
    )
    .head(10)
)
```

```python
correlation = df[
    ["budget", "revenue", "vote_average",
     "vote_count", "popularity", "roi"]
].corr()

display(
    correlation.round(2)
)
```
